```python
import json
from pathlib import Path

import pandas as pd

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 20)

# Notebooks usually run from their own folder, so find the project root.
root = Path.cwd()
if not (root / "data").exists():
    root = root.parent

files = sorted((root / "data" / "raw").glob("aqi_snapshot_*.json"))
payload = json.loads(files[-1].read_text(encoding="utf-8"))
records = [r for page in payload["pages"] for r in page["response"]["records"]]

df = pd.DataFrame(records)
print("Snapshot:", files[-1].name, "| shape:", df.shape)
df.head()
```
